```python
import pandas as pd

# Arabidopsis
# ln -s GCF_000001735.4_TAIR10.1_protein.faa 3702
# ID (3702_id.txt): GCF_000001735.4_TAIR10.1_feature_table.txt
# Soybean
# ln -s GCF_000004515.6_Glycine_max_v4.0_protein.faa 3847
# ID (3847_id.txt): GCF_000004515.6_Glycine_max_v4.0_feature_table.txt

#file1 = '3702-3702.tsv'
#file1 = '3702-3847.tsv'
#file1 = '3847-3702.tsv'
file1 = '3847-3847.tsv'
file2 = '3847_id.txt'
file3 = '3847_id.txt'
output_file = file1.rstrip('.tsv') + '_geneid.tsv'

df1 = pd.read_csv(file1, sep='\t', header=None)
df1.columns = ['product_accession_1', 'product_accession_2', 'score']
print(df1)
df2 = pd.read_csv(file2, sep='\t', dtype=str)
df2.columns = ['product_accession_1', 'geneid_1']
print(df2)
df3 = pd.read_csv(file3, sep='\t', dtype=str)
df3.columns = ['product_accession_2', 'geneid_2']
print(df3)
```

```python
df1_geneid = pd.merge(df1, df2, on='product_accession_1', how='left')
df1_geneid = pd.merge(df1_geneid, df3, on='product_accession_2', how='left')
#df1_geneid = df1_geneid.drop(df1_geneid.columns[0:3], axis=1)
#df1_geneid = df1_geneid.drop_duplicates()
print(df1_geneid)
```

```python
df1_geneid.to_csv(output_file, sep='\t', index=False)
```
